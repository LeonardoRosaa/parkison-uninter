```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("data/parkinsons.data")

columns = [
    column for column in df.columns if column not in ["spread1", "spread2", "name"]
]

x = df[[c for c in columns if c != "status"]]
y = df["status"]

healthy_df = df[df["status"] == 0]
nonhealthy_df = df[df["status"] == 1]
```

```python
columns = ["Shimmer:APQ3", "Shimmer:APQ5", "Shimmer:DDA"]

num_features = len(columns)

fig, axes = plt.subplots(num_features, num_features, figsize=(10, 10))

for i in range(num_features):
    for j in range(num_features):
        ax = axes[i, j]

        ax.scatter(
            healthy_df[columns[i]],
            healthy_df[columns[j]],
            color="blue",
            s=10,
            alpha=0.6,
            label="Healthy" if (i == 0 and j == 0) else None,
        )
        ax.scatter(
            nonhealthy_df[columns[i]],
            nonhealthy_df[columns[j]],
            color="red",
            s=10,
            alpha=0.6,
            label="Non-healthy" if (i == 0 and j == 0) else None,
        )

        if j == 0:
            ax.set_ylabel(columns[i], fontsize=14)
        if i == num_features - 1:
            ax.set_xlabel(columns[j], fontsize=14)

plt.tight_layout()
plt.show()
```
